Ejercicio 1: Clientes Duplicados y Segmentación
Contexto:
Tienes una lista de clientes con datos repetidos. Además, cuentas con dos listas: clientes  activos e inactivos.
Objetivos:
• Eliminar duplicados de la lista original.
• Identificar clientes en ambos estados.
• Encontrar clientes solo activos.
• Recorrer los clientes y clasificarlos.
Pistas:
• Usa set() para eliminar duplicados.
• Usa intersection() para encontrar coincidencias. +
• Usa difference() para encontrar diferencias.
• Usa un for con if para clasificar.

```python
# Datos de entrada
clientes_raw = ["Ana", "Luis", "Ana", "María", "Luis", "Pedro", "María", "Sofía"]
activos      = ["Ana", "Pedro", "Sofía", "Carlos"]
inactivos    = ["Luis", "María", "Jorge"]

# 1. Eliminar duplicados con set()
clientes_unicos = set(clientes_raw)
print("Clientes únicos:", clientes_unicos)

# 2. Clientes en ambos estados (intersección)
en_ambos = set(activos).intersection(set(inactivos))
print("En ambos estados:", en_ambos)

# 3. Solo activos (diferencia)
solo_activos = set(activos).difference(set(inactivos))
print("Solo activos:", solo_activos)

# 4. Clasificar con for + if
print("\n--- Clasificación de clientes únicos ---")
for cliente in sorted(clientes_unicos):
    if cliente in activos and cliente in inactivos:
        estado = "AMBOS"
    elif cliente in activos:
        estado = "ACTIVO"
    elif cliente in inactivos:
        estado = "INACTIVO"
    else:
        estado = "SIN ESTADO"
    print(f"  {cliente:10s} → {estado}")
```

Ejercicio 2: Productos y Análisis de Inventario
Contexto:
Tienes una lista de productos vendidos y otra de inventario disponible. Objetivos:
• Identificar todos los productos.
• Encontrar productos vendidos que están en inventario.
• Detectar productos que no se han vendido.
• Evaluar productos con una condición usando while.
Pistas:
• Usa union() para combinar conjuntos.
• Usa & para intersección.
• Usa - para diferencia.
• Convierte el set a lista para usar while.


```python
# --- Ejercicio 2: Productos y Análisis de Inventario ---

# Datos de entrada
vendidos   = {"laptop", "mouse", "teclado", "monitor", "audífonos"}
inventario = {"mouse", "teclado", "webcam", "hub USB", "monitor", "silla"}

# 1. Todos los productos (unión)
todos = vendidos.union(inventario)
print("Todos los productos:", todos)

# 2. Vendidos que están en inventario (intersección con &)
vendidos_en_stock = vendidos & inventario
print("Vendidos y en stock:", vendidos_en_stock)

# 3. En inventario pero NO vendidos (diferencia con -)
no_vendidos = inventario - vendidos
print("Inventario ocioso:", no_vendidos)

# 4. Evaluar con while
print("\n--- Evaluación con while ---")
lista_no_vendidos = list(no_vendidos)
i = 0
while i < len(lista_no_vendidos):
    producto = lista_no_vendidos[i]
    alerta = "REVISAR" if len(producto) > 5 else "OK"
    print(f"  [{i+1}] {producto:15s} → {alerta}")
    i += 1
```

Ejercicio 3: Estudiantes y Cursos Contexto:
Tienes estudiantes inscritos en dos cursos distintos. Objetivos:
• Encontrar estudiantes en ambos cursos. • Identificar estudiantes exclusivos por curso. • Unir todos los estudiantes.
• Clasificar estudiantes según su participación. Pistas:
• Usa & para intersección.
• Usa difference() para exclusivos.
• Usa union() para todos.
• Usa for + if para clasificar.


```python
# --- Ejercicio 3: Estudiantes y Cursos ---

# Datos de entrada
curso_python = {"Ana", "Luis", "Sofía", "Pedro", "Marta"}
curso_sql    = {"Luis", "Jorge", "Ana", "Camilo", "Marta"}

# 1. Estudiantes en ambos cursos (intersección con &)
en_ambos = curso_python & curso_sql
print("En ambos cursos:", en_ambos)

# 2. Exclusivos por curso (difference)
solo_python = curso_python.difference(curso_sql)
solo_sql    = curso_sql.difference(curso_python)
print("Solo en Python:", solo_python)
print("Solo en SQL   :", solo_sql)

# 3. Todos los estudiantes (union)
todos = curso_python.union(curso_sql)
print("Total estudiantes:", len(todos), "→", todos)

# 4. Clasificar participación con for + if
print("\n--- Clasificación por participación ---")
for estudiante in sorted(todos):
    en_py  = estudiante in curso_python
    en_sql = estudiante in curso_sql
    if en_py and en_sql:
        nivel = "AMBOS CURSOS"
    elif en_py:
        nivel = "  Python"
    else:
        nivel = "  SQL"
    print(f"  {estudiante:10s} → {nivel}")
```
